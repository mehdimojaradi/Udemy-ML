```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns   

%matplotlib inline
```

```python
df = pd.read_csv('datasets/Classified_Data.csv', index_col=0)
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(df.drop('TARGET CLASS', axis=1)) 
# desc: fit the scaler to the data (except the target class). 
# Math desc: calculate the mean and standard deviation for each feature in the dataset, which will be used for scaling.
```

```python
scaled_features = scaler.transform(df.drop('TARGET CLASS', axis=1))
# desc: transform the data using the fitted scaler.
# Math desc: apply the scaling transformation to the data, resulting in a new array where each -
# feature has been standardized (mean of 0 and standard deviation of 1).
```

### scaler.fit_transform(df.drop('TARGET CLASS', axis=1)) 
##### fit the scaler to the data and then transform it using this method in one step instead of two separate steps.

```python
df_feat = pd.DataFrame(scaled_features, columns=df.columns[:-1])
# desc: create a new DataFrame from the scaled features, using the original column names (except the target class).
df_feat.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df_feat
y = df['TARGET CLASS']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=101)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=1)
```

```python
knn.fit(X_train, y_train)
```

```python
pred = knn.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
sns.heatmap(confusion_matrix(y_test, pred), annot=True, fmt='d', cmap='Blues')
```

```python
print(classification_report(y_test, pred))
```

```python
error_rate = []
for i in range(1, 40):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)
    error_rate.append(np.mean(pred_i != y_test))
```

```python
plt.figure(figsize=(6,4))
plt.plot(range(1,40), error_rate, color='blue', linestyle='dashed', marker='o', 
        markerfacecolor='red', markersize=10)
plt.title('Error Rate vs. K Value')
plt.xlabel('K')
plt.ylabel('Error Rate')
plt.show()
# The optimal K value is the one with the lowest error rate using elbow method. In this case, it seems to be around K=17.
```

```python
knn = KNeighborsClassifier(n_neighbors=17)
knn.fit(X_train, y_train)
pred = knn.predict(X_test)
print(classification_report(y_test, pred))
sns.heatmap(confusion_matrix(y_test, pred), annot=True, fmt='d', cmap='Blues')
```
